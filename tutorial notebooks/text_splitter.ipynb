```python
text = """
Generative engine optimization (GEO) is the practice of optimizing content so that LLMs like ChatGPT and Gemini cite it as a trusted source in their responses. To do this, it’s first important to understand how LLMs source their answers.

Unlike traditional search engines that provide a list of ranked links for users to parse, LLMs synthesize answers from multiple sources. It’s looking for content that’s easy to parse and reference, and that has authority indicators showing it can be trusted.

GEO works by structuring content intentionally to make it AI-friendly in these ways, with a focus on authority, clarity, and factual accuracy.

Answer engine optimization (AEO) formats content for AI search features like Google's AI Overviews, Bing Copilot, and Perplexity's instant answers. It structures content to appear in answer-first responses that users increasingly rely on in their searches.

Instead of ranking pages, answer engines extract and present information directly within the interface. This means content must not only be optimized to rank through keywords or topical relevance, but also hyper-aligned with the questions users ask and the intent behind their searches.

AEO ensures that content can be easily extracted and delivered in these environments, with a focus on clarity, relevance, and immediate usefulness.

"""
```

```python
from langchain_text_splitters import CharacterTextSplitter

splitter = CharacterTextSplitter(
    #CharacterTextSplitter.tiktoken_encoder() -> It is a tokenization scheme used by models like: GPT-4, GPT-3.5, GPT-4o
    # it can be used to handling large documents, building RAG systems.
    # encoding_name="cl100k_base",

    separator='.',
    chunk_size=200,
    chunk_overlap=0
)

chunks = splitter.split_text(text)

len(chunks)
```

```python
for chunk in chunks:
    print(len(chunk),chunk)
    print('---')
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

splitter = RecursiveCharacterTextSplitter(
    separators=['\n\n', '\n', '.', ' '],
    chunk_size=200,
    chunk_overlap=0
)

chunks = splitter.split_text(text)
len(chunks)
```
